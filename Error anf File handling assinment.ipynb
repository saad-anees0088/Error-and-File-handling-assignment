### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4 solution code here
with open("numbers.txt", "w") as file:
    for i in range(1, 6):
        file.write(str(i) + "\n")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
with open("numbers.txt", "r") as file:
    numbers = file.readlines()

total = 0

for number in numbers:
    total += int(number)

print("Total:", total)
```

### Q9

```python
# Q9 solution code here
with open("source.txt", "w") as file:
    file.write("This is the source file.\nThis data will be copied.")
with open("source.txt", "r") as source:
    data = source.read()

with open("destination.txt", "w") as destination:
    destination.write(data)

print("File copied successfully.")
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
# Q13 solution code here
try:
    with open("non_existent_file.txt", "r") as file:
        data = file.read()
        print(data)
except FileNotFoundError:
    print("Error: File not found.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except (ZeroDivisionError, ValueError):
    print("Error: Invalid input or division by zero.")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("You entered:", number)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def age_validation(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age is valid.")

try:
    age = int(input("Enter your age: "))
    age_validation(age)
except ValueError as e:
    print("Error:", e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("Error: File not found.")

finally:
    try:
        file.close()
        print("File closed successfully.")
    except NameError:
        print("File was not opened.")
```
